```python
# 1. Counting Words in a File
# Write a function that reads the contents of a file named `document.txt` 
# and returns the number of words in the file.

def no_of_words():
    with open('document.txt','r') as file:
        content = file.read().split()
        print(f"The file contains {len(content)} words")
no_of_words()
```

```python
# # 2. Finding and Replacing Text
# Write a function that finds 
# and replaces all occurrences of a given word
# in a file named `data.txt` with another word.

def find_and_replace(old_word, new_word):
    with open('data.txt', 'r') as file:
        content = file.read()
    content=content.replace(old_word,new_word)
    with open('data.txt','w') as file:
        file.write(content)
old_word = input("Enter the word to be replaced: ")
new_word = input("Enter the new word: ")
find_and_replace(old_word,new_word)
```

```python
# 3. Reading a File in Reverse
# Write a function that reads the contents of a file named `reverse.txt` 
# and prints each line in reverse order.

def Reverse():
    with open('reverse.txt','r') as file:
        content=file.read()
        words=content.split()
        words.reverse()
        reversed_content=" ".join(words)
    with open('reverse.txt','w') as file:
        file.write(reversed_content) 

Reverse()
```

```python

# 4. Counting Lines, Words, and Characters
# Write a function that reads the contents of a file named `stats.txt`
# and returns the number of lines, words, and characters in the file.

def count_text_file(file_path):
    with open(file_path,'r')as file:
        lines=file.readlines()   #stored in list as string
        line_count=len(lines)
        word_count=sum(len(line.split())for line in lines)   #split() converts it into string
        char_count=sum(len(line)for line in lines)
    return line_count, word_count, char_count

file_path='stats.txt'
lines,words,characters=count_text_file(file_path)
print(f'Lines:{lines}\nwords:{words}\ncharcters:{characters}')

```

```python
# 15. Data Aggregation 
# Create a Pandas DataFrame with 2 columns: 'Category' and 'Value'. 
# Fill the 'Category' column with random categories ('A', 'B', 'C')
# and the 'Value' column with random integers. Group the DataFrame by 'Category' 
# and compute the sum and mean of 'Value' for each category.

data={'Category':['A','B','C'],
      'Value':[10,20,30]
      }
df=pd.DataFrame(data)
print(df.groupby('Category'))

df['Sum']=df['Value'].sum()
df['Mean']=df['Value'].mean()
df
```

```python
# 16. Create a Pandas DataFrame with 3 columns: 'Product', 'Category', and 'Sales'.
# Fill the DataFrame with random data. Group the DataFrame by 'Category' 
# and compute the total sales for each category.

data={'Category':['A','B','C','D'],
      'Value':[10,20,30,40],
      'Sales':[11,22,33,44]
 }
df=pd.DataFrame(data)
print(df.groupby('Category'))
df['Total_Sales']=df['Sales']*df['Value']
df
```

```python
# 17. Merging DataFrames
# Create two Pandas DataFrames with a common column. Merge the DataFrames using the common column.

data1={'W':[80,45,34,56,45,67],
     'X':[29,30,34,34,67,50],
     'Y':[77,10,99,20,45,78],
     'Z':[34,67,89,90,23,56]
}
df1=pd.DataFrame(data1)

data2={'X':[29,30,34,34,67,50],
       'U':[1,2,3,4,5,6],
       'V':[7,8,9,10,11,12]
}
df2=pd.DataFrame(data2)
merged=pd.merge(df1,df2,on='X')
merged
```

```python
# 18. MultiIndex DataFrame
# Create a Pandas DataFrame with a MultiIndex (hierarchical index). 
# Perform some basic indexing and slicing operations on the MultiIndex DataFrame.
index = pd.MultiIndex.from_tuples([
        ('NY',2023),('NY',2024),
        ('LA',2023),('LA',2024)],names=['city','year'])

df=pd.DataFrame({'sales':[100, 120, 90, 95]},index=index)
print(df)
print(df.loc['NY'])
print(df.loc[('LA', 2024)])
print(df.sort_index().loc['LA':'NY'])
```

```python
# 19. Create a Pandas DataFrame with MultiIndex consisting of 'Category' and 'SubCategory'.
#  Fill the DataFrame with random data and compute the sum of values for each 'Category' and 'SubCategory'.

index=pd.MultiIndex.from_tuples(
    [('Electronics','A'),('Electronics','B'),
     ('Clothing','A'),('Clothing','B')],
    names=['Category','SubCategory']
)

df=pd.DataFrame({'Value':[50,30,20,40]},index=index)
print(df.groupby(level='Category').sum())
print(df.groupby(level='SubCategory').sum())
```

```python
# 21. Create a Pandas DataFrame with 3 columns and 6 rows filled with random integers.
#  Apply a lambda function to create a new column that is the sum of the existing columns.

df=pd.DataFrame({'A':[1,2,3],'B':[4,5,6],'C':[7,8,9]})
df['Sum']=df.apply(lambda row:row['A']+row['B']+row['C'],axis=1)
print(df)
```
